# Olist SQL EDA scratchpad
Prototype database-side audits here before moving stable SQL into `sql/` and `scripts/`. No dataframe libraries are used.

```python
from config import DB_CONFIG
import psycopg2
conn = psycopg2.connect(**DB_CONFIG)
cur = conn.cursor()
```

```python
cur.execute("""
SELECT order_status, COUNT(*) AS orders,
       COUNT(*) FILTER (WHERE order_delivered_customer_date IS NULL) AS missing_delivery_date
FROM public.orders
GROUP BY order_status ORDER BY orders DESC
""")
cur.fetchall()
```

```python
cur.close()
conn.close()
```
